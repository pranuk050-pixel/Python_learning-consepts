```python
### Preprocessing pipeline text data
  ### Source :text data, s3 bucket,azure blob storage, mongo db, vector databases
  ### lower case or upper case
  ### stopwords removal
  ### special character removal
    ###  conside  emoticons, emoji
  ### abbrevations
  ### lemmatization
  ### stemming
  ### <h1><br> Tag removal
  ### spell checker library : correct the spellings
  ### expand contractions don`t do not can`t cannot
  ### remove hashtags (This is specifically for social media data)
  ### remove duplicate
  ### text normalization : USA, U.S.A , united states
```

```python
# 1. Lowercase
text = text.lower()
print(text)

# 2. Remove URL
text = re.sub(r"https?://\S+", "", text)
print(text)

# 3. Remove mentions
text = re.sub(r"@\w+", "", text)
print(text)

# 4. Remove hashtag symbol
text = re.sub(r"#", "", text)
print(text)

# 5. Remove numbers
text = re.sub(r"\d+", "", text)
print(text)

# 6. Remove punctuation / special characters
text = re.sub(r"[^\w\s]", "", text)
print(text)

# 7. Remove extra spaces
text = re.sub(r"\s+", " ", text).strip()
print(text)

# Final Output:
# i love python visit python
```
