```python
### Meta characters
```

```python
^ : This checks wheather a text is starts with a pattern
$ : This checks wheather a text is ends with a pattern
```

```python
text = " sciendatace"
pattern = "^data"
print(re.findall(pattern, text))
```

```python
text = " sciendatace"
pattern = "data$"
print(re.findall(pattern, text))
```

```python
text = "7sciendatace"
pattern = "^[A-Za-z]"
print(re.findall(pattern, text))
```

```python
{} : Exact number of occurances / restricted number of occurance
| :
```

```python
import re

text = "a aa aaa aaaa"
pattern = "a{3}"

print(re.findall(pattern, text))
```

```python
import re

text = "1 12 123 45 678"
pattern = r"\d{2}"

print(re.findall(pattern, text))
```

```python
import re

text = "hee heXXo hee12o heABheeeeeeo"
pattern = r"hen{2}"

print(re.findall(pattern, text))
```
