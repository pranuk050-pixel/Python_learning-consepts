```python
import pandas as pd

try:
    df = pd.read_csv("sales.csv")

    revenue = float(df["revenue"].iloc[0])
    customers = int(df["customers"].iloc[0])

    revenue_per_customer = revenue / customers

    print("Revenue per Customer:", revenue_per_customer)

except FileNotFoundError:
    print("Error: Input file not found")

except KeyError as e:
    print("Error: Required column missing:", e)

except ValueError:
    print("Error: Invalid numeric value")

except ZeroDivisionError:
    print("Error: Number of customers cannot be zero")

except IndexError:
    print("Error: Dataset contains no records")

except Exception as e:
    print("Unexpected Error:", e)

else:
    print("ETL pipeline completed successfully")

finally:
    print("Closing pipeline resources")
```
