```python
class Product:
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price

    def show_product(self):
        print("Product Name:", self.product_name)
        print("Price:", self.price)

    def calculate_discount(self, discount):
        discount_amount = self.price * discount / 100
        final_price = self.price - discount_amount
        print("Final Price:", final_price)


class Electronics(Product):
    def __init__(self, product_name, price, warranty):
        super().__init__(product_name, price)
        self.warranty = warranty

    def show_warranty(self):
        print("Warranty:", self.warranty)

    def extend_warranty(self, years):
        print("Extended Warranty:", years, "Years")


e1 = Electronics("Laptop", 50000, "2 Years")

e1.show_warranty()
e1.extend_warranty(10)
e1.show_product()
e1.calculate_discount(3)
```

```python
import pandas as pd
import matplotlib.pyplot as plt


# Parent Class
class Data:
    def __init__(self):
        self.data = {
            "Product": ["Laptop", "Mobile", "TV", "Tablet", "Watch"],
            "Sales": [50000, 35000, 40000, 20000, 15000]
        }

        self.df = pd.DataFrame(self.data)

    def show_data(self):
        print("Sales Data")
        print(self.df)


# Child Class
class DataAnalysis(Data):
    def __init__(self):
        Data.__init__(self)

    def show_total_sales(self):
        total = self.df["Sales"].sum()
        print("Total Sales:", total)

    def show_average_sales(self):
        average = self.df["Sales"].mean()
        print("Average Sales:", average)


# Grandchild Class
class DataVisualization(DataAnalysis):
    def __init__(self):
        DataAnalysis.__init__(self)

    def bar_chart(self):
        plt.bar(self.df["Product"], self.df["Sales"])
        plt.title("Product Sales")
        plt.xlabel("Product")
        plt.ylabel("Sales")
        plt.show()

    def line_chart(self):
        plt.plot(self.df["Product"], self.df["Sales"], marker="o")
        plt.title("Product Sales Trend")
        plt.xlabel("Product")
        plt.ylabel("Sales")
        plt.show()


v1 = DataVisualization()

v1.show_data()
v1.show_total_sales()
v1.show_average_sales()
v1.bar_chart()
v1.line_chart()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


# Parent Class
class SalesData:
    def __init__(self):
        self.data = {
            "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
            "Revenue": [25000, 32000, 28000, 40000, 45000]
        }

        self.df = pd.DataFrame(self.data)

    def show_data(self):
        print("Monthly Sales Data")
        print(self.df)


# Child Class
class SalesAnalysis(SalesData):
    def __init__(self):
        SalesData.__init__(self)

    def total_revenue(self):
        total = self.df["Revenue"].sum()
        print("Total Revenue:", total)

    def maximum_revenue(self):
        maximum = self.df["Revenue"].max()
        print("Maximum Revenue:", maximum)


# Grandchild Class
class SalesVisualization(SalesAnalysis):
    def __init__(self):
        SalesAnalysis.__init__(self)

    def bar_chart(self):
        plt.bar(self.df["Month"], self.df["Revenue"])
        plt.title("Monthly Revenue")
        plt.xlabel("Month")
        plt.ylabel("Revenue")
        plt.show()

    def line_chart(self):
        plt.plot(
            self.df["Month"],
            self.df["Revenue"],
            marker="o"
        )
        plt.title("Revenue Trend")
        plt.xlabel("Month")
        plt.ylabel("Revenue")
        plt.show()


v1 = SalesVisualization()

v1.show_data()
v1.total_revenue()
v1.maximum_revenue()
v1.bar_chart()
v1.line_chart()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def show_product(self):
        print("Product:", self.name)
        print("Price:", self.price)


class Brand:
    def __init__(self, brand):
        self.brand = brand

    def show_brand(self):
        print("Brand:", self.brand)


class Electronics(Product, Brand):
    def __init__(self, name, price, brand):
        Product.__init__(self, name, price)
        Brand.__init__(self, brand)

    def show_details(self):
        print("Electronics Product Details")


e1 = Electronics("Laptop", 75000, "Dell")

e1.show_product()
e1.show_brand()
e1.show_details()
```

```python
#hands'on
class Customer:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def show_customer(self):
        print("Customer:", self.name)
        print("Phone:", self.phone)


class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def show_account(self):
        print("Account Number:", self.account_number)
        print("Balance:", self.balance)


class SavingsAccount(Customer, Account):
    def __init__(self, name, phone, account_number, balance):
        Customer.__init__(self, name, phone)
        Account.__init__(self, account_number, balance)

    def show_details(self):
        print("Savings Account Active")


a1 = SavingsAccount("Rahul", "9876543210", 1001, 50000)

a1.show_customer()
a1.show_account()
a1.show_details()
```

```python
#hands'on
class DataSource:
    def __init__(self, file_name):
        self.file_name = file_name

    def load_data(self):
        print("File Name:", self.file_name)
        print("Data Loaded Successfully")


class Algorithm:
    def __init__(self, model_name):
        self.model_name = model_name

    def show_model(self):
        print("Model:", self.model_name)


class DataScientist(DataSource, Algorithm):
    def __init__(self, file_name, model_name, accuracy):
        DataSource.__init__(self, file_name)
        Algorithm.__init__(self, model_name)
        self.accuracy = accuracy

    def show_accuracy(self):
        print("Accuracy:", self.accuracy)


ds1 = DataScientist(
    "customers.csv",
    "Random Forest",
    94.5
)

ds1.load_data()
ds1.show_model()
ds1.show_accuracy()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def show_product(self):
        print("Product:", self.name)
        print("Price:", self.price)


class Laptop(Product):
    def __init__(self, name, price, ram):
        Product.__init__(self, name, price)
        self.ram = ram

    def show_ram(self):
        print("RAM:", self.ram)


class Mobile(Product):
    def __init__(self, name, price, camera):
        Product.__init__(self, name, price)
        self.camera = camera

    def show_camera(self):
        print("Camera:", self.camera)


class TV(Product):
    def __init__(self, name, price, size):
        Product.__init__(self, name, price)
        self.size = size

    def show_size(self):
        print("TV Size:", self.size)


l1 = Laptop("Dell Laptop", 75000, "16 GB")
m1 = Mobile("Samsung Mobile", 45000, "50 MP")
t1 = TV("Sony TV", 60000, "55 Inch")


l1.show_product()
l1.show_ram()

m1.show_product()
m1.show_camera()

t1.show_product()
t1.show_size()
```

```python
class BankAccount:
    def __init__(self, customer_name, balance):
        self.customer_name = customer_name
        self.balance = balance

    def show_account(self):
        print("Customer Name:", self.customer_name)
        print("Balance:", self.balance)


class SavingsAccount(BankAccount):
    def __init__(self, customer_name, balance, interest_rate):
        BankAccount.__init__(self, customer_name, balance)
        self.interest_rate = interest_rate

    def show_interest(self):
        print("Interest Rate:", self.interest_rate)


class CurrentAccount(BankAccount):
    def __init__(self, customer_name, balance, overdraft_limit):
        BankAccount.__init__(self, customer_name, balance)
        self.overdraft_limit = overdraft_limit

    def show_overdraft(self):
        print("Overdraft Limit:", self.overdraft_limit)


class LoanAccount(BankAccount):
    def __init__(self, customer_name, balance, loan_amount):
        BankAccount.__init__(self, customer_name, balance)
        self.loan_amount = loan_amount

    def show_loan(self):
        print("Loan Amount:", self.loan_amount)


s1 = SavingsAccount("Rahul", 50000, 6.5)
c1 = CurrentAccount("Ravi", 100000, 25000)
l1 = LoanAccount("Priya", 30000, 500000)

s1.show_account()
s1.show_interest()

c1.show_account()
c1.show_overdraft()

l1.show_account()
l1.show_loan()
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("Name:", self.name)


class Employee(Person):
    def __init__(self, name, emp_id):
        Person.__init__(self, name)
        self.emp_id = emp_id

    def show_employee(self):
        print("Employee ID:", self.emp_id)


class Student(Person):
    def __init__(self, name, course):
        Person.__init__(self, name)
        self.course = course

    def show_course(self):
        print("Course:", self.course)


class Intern(Employee, Student):
    def __init__(self, name, emp_id, course, duration):
        Employee.__init__(self, name, emp_id)
        Student.__init__(self, name, course)
        self.duration = duration

    def show_duration(self):
        print("Internship Duration:", self.duration)


i1 = Intern("Rahul", 101, "Data Science", "6 Months")

i1.show_name()
i1.show_employee()
i1.show_course()
i1.show_duration()
```

```python
class DataSource:
    def __init__(self, source_name):
        self.source_name = source_name

    def show_source(self):
        print("Data Source:", self.source_name)


class BatchProcessing(DataSource):
    def __init__(self, source_name, batch_size):
        DataSource.__init__(self, source_name)
        self.batch_size = batch_size

    def show_batch(self):
        print("Batch Size:", self.batch_size)


class StreamProcessing(DataSource):
    def __init__(self, source_name, interval):
        DataSource.__init__(self, source_name)
        self.interval = interval

    def show_stream(self):
        print("Streaming Interval:", self.interval)


class DataPipeline(BatchProcessing, StreamProcessing):
    def __init__(self, source_name, batch_size, interval, pipeline_name):
        BatchProcessing.__init__(self, source_name, batch_size)
        StreamProcessing.__init__(self, source_name, interval)
        self.pipeline_name = pipeline_name

    def show_pipeline(self):
        print("Pipeline:", self.pipeline_name)


p1 = DataPipeline(
    "Sales Database",
    10000,
    "5 Minutes",
    "Sales Pipeline"
)

p1.show_source()
p1.show_batch()
p1.show_stream()
p1.show_pipeline()
```
