```python
class Student:
    def __init__(self, name, marks, course):
        self.name = name
        self.marks = marks
        self.course = course

    def display(self):
        print("**********************")
        print("Student Name:", self.name)
        print("Marks:", self.marks)
        print("Course:", self.course)
        print("**********************")

    def result(self):
        if self.marks >= 40:
            print("Pass")
        else:
            print("Fail")

    def add_bonus(self, bonus):
        self.marks = self.marks + bonus
        print("New Marks:", self.marks)


s1 = Student("Rahul", 75, "Python")
s1.display()
s1.result()
s1.add_bonus(5)
```

```python
class Product:
    def __init__(self, name, price, quantity, category):
        self.name = name
        self.price = price
        self.quantity = quantity
        self.category = category

    def display(self):
        print("**********************")
        print("Product:", self.name)
        print("Category:", self.category)
        print("Price:", self.price)
        print("Stock:", self.quantity)
        print("**********************")

    def total_stock_value(self):
        value = self.price * self.quantity
        print("Stock Value:", value)

    def sell_product(self, units):
        if units <= self.quantity:
            self.quantity = self.quantity - units
            print(units, "units sold")
            print("Remaining Stock:", self.quantity)
        else:
            print("Insufficient Stock")

    def restock(self, units):
        self.quantity = self.quantity + units
        print("Updated Stock:", self.quantity)

    def apply_discount(self, discount):
        self.price = self.price - (self.price * discount / 100)
        print("Discounted Price:", self.price)


p1 = Product("Laptop", 10,60000, "Electronics")

p1.display()
p1.total_stock_value()
p1.sell_product(3)
p1.restock(5)
p1.apply_discount(10)
```

```python
import pandas as pd

class DatasetAnalyzer:

    def __init__(self, file_path):
        self.file_path = file_path
        if "xls" in file_path:
          self.df = pd.read_excel(file_path)
        elif "csv" in file_path:
          self.df = pd.read_csv(file_path)

    def display(self):
        print(self.df.head())

    def shape(self):
        print("Rows:", self.df.shape[0])
        print("Columns:", self.df.shape[1])

    def missing_values(self):
        print(self.df.isnull().sum())

    def statistics(self):
        print(self.df.describe())

```

```python
import pandas as pd

class DataCleaner:

    def __init__(self, file_path):
        self.df = pd.read_csv(file_path)

    def remove_duplicates(self):
        self.df = self.df.drop_duplicates()
        print("Duplicates Removed")

    def fill_missing(self, column):
        self.df[column] = self.df[column].fillna(
            self.df[column].mean()
        )
        print("Missing Values Filled")

    def remove_column(self, column):
        self.df = self.df.drop(columns=[column])
        print(column, "removed")

    def save_data(self, output_file):
        self.df.to_csv(output_file, index=False)
        print("Saved:", output_file)


cleaner = DataCleaner("customers.csv")

cleaner.remove_duplicates()
cleaner.fill_missing("age")
cleaner.remove_column("customer_id")
cleaner.save_data("clean_customers.csv")
```

```python
import pandas as pd

class SalesAnalyzer:

    def __init__(self, file_path):
        self.df = pd.read_csv(file_path)

    def total_sales(self):
        total = self.df["sales"].sum()
        print("Total Sales:", total)

    def average_sales(self):
        average = self.df["sales"].mean()
        print("Average Sales:", average)

    def top_product(self):
        result = self.df.groupby("product")["sales"].sum()
        print(result.sort_values(ascending=False).head(1))

    def region_sales(self):
        result = self.df.groupby("region")["sales"].sum()
        print(result)


sales = SalesAnalyzer("sales.csv")

sales.total_sales()
sales.average_sales()
sales.top_product()
sales.region_sales()
```

```python
import pandas as pd

class ETLPipeline:

    def __init__(self, source, destination):
        self.source = source
        self.destination = destination
        self.df = None

    def extract(self):
        self.df = pd.read_csv(self.source)
        print("Data Extracted")

    def transform(self):
        self.df = self.df.drop_duplicates()
        self.df = self.df.dropna()

        print("Data Transformed")

    def load(self):
        self.df.to_csv(
            self.destination,
            index=False
        )

        print("Data Loaded")


pipeline = ETLPipeline(
    "raw_sales.csv",
    "clean_sales.csv"
)

pipeline.extract()
pipeline.transform()
pipeline.load()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

class DataVisualizer:

    # Constructor
    def __init__(self, file_path):
        self.file_path = file_path
        self.df = pd.read_csv(file_path)

        print("Dataset Loaded Successfully")
        print("Shape:", self.df.shape)


    # Method 1 - Line Chart
    def line_chart(self, x, y):
        plt.plot(self.df[x], self.df[y])

        plt.xlabel(x)
        plt.ylabel(y)
        plt.title("Line Chart")

        plt.show()


    # Method 2 - Bar Chart
    def bar_chart(self, x, y):
        plt.bar(self.df[x], self.df[y])

        plt.xlabel(x)
        plt.ylabel(y)
        plt.title("Bar Chart")

        plt.show()


    # Method 3 - Scatter Plot
    def scatter_plot(self, x, y):
        plt.scatter(self.df[x], self.df[y])

        plt.xlabel(x)
        plt.ylabel(y)
        plt.title("Scatter Plot")

        plt.show()


    # Method 4 - Histogram
    def histogram(self, column):
        plt.hist(self.df[column])

        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.title("Histogram")

        plt.show()


    # Method 5 - Box Plot
    def box_plot(self, column):
        plt.boxplot(self.df[column])

        plt.ylabel(column)
        plt.title("Box Plot")

        plt.show()


    # Method 6 - Pie Chart
    def pie_chart(self, column):
        data = self.df[column].value_counts()

        plt.pie(
            data.values,
            labels=data.index,
            autopct="%1.1f%%"
        )

        plt.title("Pie Chart")

        plt.show()
```

```python
import pandas as pd
import sqlite3
import mysql.connector
import psycopg2
import pyodbc
import oracledb
```

```python


class DatabaseConnector:

    # Constructor
    def __init__(self, host, username, password, database):
        self.host = host
        self.username = username
        self.password = password
        self.database = database


    # Method 1 - MySQL
    def connect_mysql(self):

        connection = mysql.connector.connect(
            host=self.host,
            user=self.username,
            password=self.password,
            database=self.database
        )

        print("MySQL Connected Successfully")

        return connection


    # Method 2 - PostgreSQL
    def connect_postgresql(self):

        connection = psycopg2.connect(
            host=self.host,
            user=self.username,
            password=self.password,
            dbname=self.database
        )

        print("PostgreSQL Connected Successfully")

        return connection


    # Method 3 - SQL Server
    def connect_sqlserver(self):

        connection = pyodbc.connect(
            "DRIVER={ODBC Driver 18 for SQL Server};"
            f"SERVER={self.host};"
            f"DATABASE={self.database};"
            f"UID={self.username};"
            f"PWD={self.password};"
            "TrustServerCertificate=yes;"
        )

        print("SQL Server Connected Successfully")

        return connection


    # Method 4 - Oracle
    def connect_oracle(self):

        connection = oracledb.connect(
            user=self.username,
            password=self.password,
            dsn=f"{self.host}:1521/{self.database}"
        )

        print("Oracle Connected Successfully")

        return connection


    # Method 5 - SQLite
    def connect_sqlite(self):

        connection = sqlite3.connect(self.database)

        print("SQLite Connected Successfully")

        return connection
```
