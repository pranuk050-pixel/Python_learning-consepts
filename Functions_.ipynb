````python
name="Data Engineering" ### global varaible
def outer_function():
  name="Data science"
  def inner_function():
    nonlocal name
    name="Artificial Intelligence"
  inner_function()
  print(name)
outer_function()
````

````python
def outer():

    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()

    print(x)

outer()
````

````python
x = 100      # Global variable
def outer():
    y = 200   # Enclosing variable
    def inner():
        global x
        nonlocal y
        x = x + 50
        y = y + 50
        print("Inside Inner")
        print("Global x :", x)
        print("Enclosing y :", y)
    inner()
    print("Inside Outer")
    print("y =", y)
outer()
print("Outside")
print("x =", x)
````

````python
count = 0
def company():
    employees = 10
    def hire():
        global count
        nonlocal employees
        count += 1
        employees += 5
        print("Employees:", employees)
        print("Global Count:", count)
    hire()
    print("Final Employees:", employees)
company()
print("Total Companies Processed:", count)
````

````python
bank_balance = 10000
def account():
    savings = 5000
    def deposit():
        global bank_balance
        nonlocal savings
        bank_balance += 2000
        savings += 1000
        print("Savings =", savings)
        print("Bank Balance =", bank_balance)
    deposit()
    print("Account Savings =", savings)
account()
print("Global Balance =", bank_balance)
````

````python
x = 10
def outer():
    y = 20
    def middle():
        z = 30
        def inner():
            global x
            nonlocal y
            nonlocal z
            x += 1
            y += 2
            z += 3
            print("Inner")
            print(x, y, z)
        inner()
        print("Middle:", z)
    middle()
    print("Outer:", y)
outer()
print("Global:", x)
````

````python
### global : varaibles declared outside all functions : globa
### nonlocal: varaible declared in the nearest to the enclosing functions, nonlocal
### LEG : local, enclosing brackets, global
````

# Population vs Sample

|Feature|Population|Sample|
|--------|----------|------|
|Definition|Entire dataset|Subset of dataset|
|Size|Large|Smaller|
|Cost|High|Low|
|Time|More|Less|
|Purpose|Complete study|Estimate population characteristics|
